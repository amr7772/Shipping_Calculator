```python
#Free Weekend Delivery Check
```

```python
Order_value = float(input("Please enter the order value: "))

week_day = input("Please enter the day of purchase: ")

if week_day == "Saturday" or week_day == "Sunday":
    print("It's free delivery")

else:
    print(f"The delivery fee for the order you made on {week_day} with ${Order_value} is $10")
```

```python
Order_value = float(input("Please enter the order value: "))

week_day = input("Please enter the day of purchase: ")

if Order_value >= 1000:
    print("It's free delivery")

else:
    print(f"The delivery fee for the order you made on {week_day} with ${Order_value} is $10")
```
